```python
import os
import cv2
from ultralytics import YOLO
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
YOLO_model = "/content/drive/MyDrive/TA134!/Case_Data_Train_Gabungan/Training/weights/best.pt"
model = YOLO(YOLO_model)

video_input = "/content/drive/MyDrive/TA134!/malam2.mp4"
output_folder = "/content/drive/MyDrive/TA134!/Uji_Video/Hasil_Deteksi_Maalam"

os.makedirs(output_folder, exist_ok=True)

video_output = os.path.join(
    output_folder,
    "hasil_deteksi_maalam.mp4"
)

cap = cv2.VideoCapture(video_input)

if not cap.isOpened():
    print("Video gagal dibuka")
    exit()

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps_video = cap.get(cv2.CAP_PROP_FPS)

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

out = cv2.VideoWriter(
    video_output,
    fourcc,
    fps_video,
    (width, height)
)

frame_num = 0

print("Memulai deteksi kendaraan pada video")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    results = model.predict(
        source=frame,
        conf=0.5,
        imgsz=640,
        verbose=False
    )

    annotated_frame = results[0].plot()

    out.write(annotated_frame)
    frame_num += 1

cap.release()
out.release()

print("=" * 40)
print("PROSES SELESAI")
print("=" * 40)
print(f"Total frame diproses : {frame_num}")
print(f"Video tersimpan di   : {video_output}")
print("=" * 40)
```
